```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.metrics import mean_squared_error

# ------------------------------
# Load dataset
# ------------------------------
df = pd.read_csv("real_data.csv")    # Use your dataset
X = df[['X']].values
y = df['y'].values

# ------------------------------
# Linear Regression
# ------------------------------
lin = LinearRegression()
lin.fit(X, y)

# manual predicted values = y_pred_linear
y_pred_linear = lin.predict(X)

# manual MSE
mse_linear = mean_squared_error(y, y_pred_linear)

print("\n=== Linear Regression ===")
print("Coefficient (m):", lin.coef_[0])
print("Intercept (c):", lin.intercept_)
print("Predicted Values:", y_pred_linear)
print("MSE:", mse_linear)

# ------------------------------
# Ridge Regression
# ------------------------------
ridge = Ridge(alpha=1.0)
ridge.fit(X, y)

y_pred_ridge = ridge.predict(X)
mse_ridge = mean_squared_error(y, y_pred_ridge)

print("\n=== Ridge Regression ===")
print("Coefficient (m):", ridge.coef_[0])
print("Intercept (c):", ridge.intercept_)
print("Predicted Values:", y_pred_ridge)
print("MSE:", mse_ridge)

# ------------------------------
# Lasso Regression
# ------------------------------
lasso = Lasso(alpha=0.1)
lasso.fit(X, y)

y_pred_lasso = lasso.predict(X)
mse_lasso = mean_squared_error(y, y_pred_lasso)

print("\n=== Lasso Regression ===")
print("Coefficient (m):", lasso.coef_[0])
print("Intercept (c):", lasso.intercept_)
print("Predicted Values:", y_pred_lasso)
print("MSE:", mse_lasso)

# ------------------------------
# Which model is best?
# ------------------------------
print("\n=== Model Comparison (Based on MSE) ===")
print("Linear MSE:", mse_linear)
print("Ridge  MSE:", mse_ridge)
print("Lasso  MSE:", mse_lasso)

best_model = min(
    [("Linear", mse_linear), ("Ridge", mse_ridge), ("Lasso", mse_lasso)],
    key=lambda x: x[1]
)

print("\nBest Model:", best_model[0])
```
